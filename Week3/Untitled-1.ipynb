```python
# 17 Create a class Product with attributes name and price. Use a constructor to initialize them. Add a method discount() that calculates a 10% discount and displays the final price.
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def discount(self):
        discounted_price = self.price * 0.9
        print(f"Product Name: {self.name}")
        print(f"Original Price: ${self.price}")
        print(f"Final Price: ${self.price-discounted_price} with a discount ${discounted_price}")
p=Product("Smartphone", 500)
p.discount()        
```
